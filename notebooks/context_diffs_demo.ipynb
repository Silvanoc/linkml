# Demonstrate differences in JSON-LD contexts

The [JSON-LD Generator](https://github.com/linkml/linkml/blob/main/linkml/generators/jsonldgen.py) and the [Context Generator](https://github.com/linkml/linkml/blob/main/linkml/generators/jsonldcontextgen.py) somehow yield different contexts. This notebook demonstrates it.

```python
from linkml.generators.jsonldgen import JSONLDGenerator
from linkml.generators.jsonldcontextgen import ContextGenerator
from pathlib import Path
import json
from rdflib import Graph, compare
from pprint import pprint

schema_yaml_path = "../tests/test_scripts/input/simple_uri_test-monolithic.yaml"
```

## Generate schema JSON-LD

```python
jsonld_gen = JSONLDGenerator(schema_yaml_path)
tmp_dir = Path.cwd() / "tmp"
direct_jsonld = json.loads(jsonld_gen.serialize())
json.dump(direct_jsonld, open(str(tmp_dir / "simple_uri_test-monolithic.ld.json"), "w"), indent=2)
```

### Print context

```python
pprint(direct_jsonld["@context"], indent=2)
```

## Generate contexts

Generate the contexts with all 4 different combinations of the flags `model` and `prefixes`.

### Model=True, Prefixes=True

```python
model = True
prefixes = True
context_1 = json.loads(ContextGenerator(schema_yaml_path, model=model, prefixes=prefixes).serialize(model=model, prefixes=prefixes))
```

#### Print context

```python
pprint(context_1["@context"])
```

### Model=True, Prefixes=False

```python
model = True
prefixes = False
context_2 = json.loads(ContextGenerator(schema_yaml_path, model=model, prefixes=prefixes).serialize(model=model, prefixes=prefixes))
```

#### Print context

```python
pprint(context_2["@context"])
```

### Model=False, Prefixes=True

```python
model = False
prefixes = True
context_3 = json.loads(ContextGenerator(schema_yaml_path, model=model, prefixes=prefixes).serialize(model=model, prefixes=prefixes))
```

#### Print context

```python
pprint(context_3["@context"])
```

### Model=False, Prefixes=False

```python
model = False
prefixes = False
context_4 = json.loads(ContextGenerator(schema_yaml_path, model=model, prefixes=prefixes).serialize(model=model, prefixes=prefixes))
```

#### Print context

```python
pprint(context_4["@context"])
```

## Expand schema via RDF transformations

### Expand the schema with the JSON-LD context from JSONLDGenerator

```python
g_direct = Graph().parse(data=direct_jsonld, format="json-ld")
g_direct_expanded = json.loads(g_direct.serialize(format="json-ld", indent=2))
```

#### Print overview

```python
for item in g_direct_expanded:
    if  ("@type" in item.keys() and item["@type"] == ['https://w3id.org/linkml/SlotDefinition']) or \
        ("@type" in item.keys() and item["@type"] == ['https://w3id.org/linkml/ClassDefinition']) or \
        ("@type" in item.keys() and item["@type"] == ['https://w3id.org/linkml/TypeDefinition']) or \
        ("@type" in item.keys() and item["@type"] == ['https://w3id.org/linkml/SchemaDefinition']) or \
        ("@type" in item.keys() and item["@type"] == ['https://w3id.org/linkml/Annotation']):
        print(f"🟢 {item["@id"]} is a {item["@type"]}")
    elif "http://www.w3.org/ns/shacl#namespace" in item.keys():
        print(f"🟢 {item["@id"]} is the Namespace {item["http://www.w3.org/ns/shacl#namespace"]}")
    else:
        if "@type" in item.keys():
            print(f"☢️ {item["@id"]} is a {item["@type"]}")
        else:
            print(item)
```

#### Print full result

```python
pprint(g_direct_expanded)
```
